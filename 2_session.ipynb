# Operators in Python


```python
# Arithmetic operators
print(5+6)
print(5-4)
print(5*2)
print(5/2) #deimal division
print(5//2) #integer division
print(5%2) #modulus
print(5**2) #exponentiation
```

```python
# Relational operators
print(5>2)
print(5<2)
print(5>=2)
print(5<=2)
print(5==2)
print(5!=2)
```

```python
# Logical operators
print(5>2 and 6>5)
print(5<2 or 6<5)
print(not(2>5))
print(1 and 1)
print( 1 and 0)
print(0 and 1)
print(0 and 0)
print(1 or 1)
print(0 or 1)
print(1 or 0)
print(0 or 0)
print(not 0)
print(not 1)
```

```python
# Bitwise operators : AND(&), OR(|), NOT(~), XOR(^), Left Shift(<<), Right Shift(>>) and works on bits
# example: a=5(0101), b=3(0011)
a = 5
b = 3
print(a & b)  # AND : 0101 & 0011 = 0001 = 1
print(a | b)  # OR  : 0101 | 0011 = 0111 = 7
print(a ^ b)  # XOR : 0101 ^ 0011 = 0110 = 6
print(~a)     # NOT : ~0101 = 1010 = -6 (two's complement)
print(a << 1) # Left Shift : 0101 << 1 = 1010 = 10
print(a >> 1) # Right Shift: 0101 >> 1 = 0010 = 2   
```

```python
# Assignment operators
a = 5
a += 2  # a = a + 2
print(a)
a -= 2  # a = a - 2
print(a)
a *= 2  # a = a * 2
print(a)
a /= 2  # a = a / 2
print(a)
a //= 2 # a = a // 2
print(a)
a %= 2  # a = a % 2
print(a)
a **= 2 # a = a ** 2
print(a)
```

```python
# Memebership operators : defines if a value is present in a sequence
list1 = [1,2,3,4,5]
print(6 in list1)
print(3 in list1)
print(6 not in list1)
print(3 not in list1)
```

```python
# program - Find the sum of a 3 digit number entered by the user
num = int(input("Enter any digit number: "))
digit1 = num % 10
num = num//10
digit2 = num % 10
num = num//10
digit3 = num % 10
sum = digit1 + digit2 + digit3
print("The sum of the digits is:", sum)
```

## IF ELSE in Python

```python
# login program and indentation
email = input("Enter your email id: ")
password = input("Enter your password: ")
if(email == "prabhat@gmail.com" and password == "12345"):
    print("Login Successful")
else:
    print("Login Failed")
```

```python
# login program and indentation
email = input("Enter your email id: ")
password = input("Enter your password: ")
if(email == "prabhat@gmail.com" and password == "12345"):
    print("Login Successful")

elif(email != "prabhat@gmail.com" and password == "12345"):
    print("Email id is incorrect")
    email = input("Enter your email id again: ")
    if(email=="prabhat@gmail.com"):
        print("Login Successful")
    else:
        print("Login Failed")
elif(password != "12345" and email == "prabhat@gmail.com"):
    print("Password is incorrect")
    password = input("Enter your password again: ")
    if(password=="12345"):
        print("Login Successful")
    else:
        print("Login Failed")
else:
    print("Login Failed")
```

```python
# find the min number among three numbers
num1,num2,num3 = map(int,input("Enter 1st , 2nd and 3rd number: ").split())
if(num1<=num2 and num1<=num3):
    print("The minimum number is:", num1)
elif(num2<=num1 and num2<=num3):
    print("The minimum number is:", num2)
else:
    print("The minimum number is:", num3)
```

```python
# Menu driven Calculator
fnum = int(input("Enter first number: "))
snum = int(input("Enter second number: "))
op = input("Enter operator (+, -, *, /): ")
if op == '+':
    print("The sum is:", fnum + snum)
elif op == '-':
    print("The difference is:", fnum - snum)
elif op == '*':
    print("The product is:", fnum * snum)
elif op == '/':
    print("The quotient is:", fnum / snum)
else:
    print("Invalid operator")
```

# Loops in Python

```python
# while loop in Python
# take a number from the user and print thier table
num = int(input("Enter a number to print its table: "))
i = 1
while i <= 10:
    print(f"{num} * {i} = {num*i}")
    i += 1
```

```python
# while loop with else
x = 1
while x<3:
    print(x)
    x += 1
else:
    print("Loop ended")
```

```python
# Gussing Game
import random
number_to_guess = random.randint(1,100)
user_guess = int(input("Enter your guess (1-100): "))
attempts = 1
while user_guess != number_to_guess:
     
     
     if user_guess < number_to_guess:
        print("Too low! Try again.")
     else:
        print("Too high! Try again.")
     user_guess = int(input("Enter your guess (1-100): "))
     attempts += 1
else:
    print(f"Congratulations! You've guessed the number {number_to_guess} in {attempts} attempts.")
        
```
